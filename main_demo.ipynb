```python
import pandas as pd
import matplotlib.pyplot as plt

dati = {
    "comune": [
        "Comune A",
        "Comune B",
        "Comune C",
        "Comune D",
        "Comune E",
        "Comune F"
    ],
    "incidenti_per_100k": [100, 120, 350, 380, 700, 750],
    "incidenti_per_km2": [1, 2, 6, 7, 13, 14]
}

df = pd.DataFrame(dati)

plt.scatter(
    df["incidenti_per_100k"],
    df["incidenti_per_km2"]
)

plt.xlabel("Incidenti ogni 100.000 abitanti")
plt.ylabel("Incidenti per km²")
plt.title("Distribuzione degli incidenti per comune")

plt.show()
```
